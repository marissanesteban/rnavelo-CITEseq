```python
import gseapy as gp

# different gene sets
gene_sets = [
    "GO_Biological_Process_2023",
    "MSigDB_Hallmark_2020",
    "ChEA_2022",
    "ENCODE_TF_ChIP-seq_2015",
    "Reactome_2022"
]

TOP_N = 200 
basal2_genes = rank_df["gene"].head(TOP_N).tolist()

# GO enrichment (Enrichr library via gseapy)
enr = gp.enrichr(
    gene_list=basal2_genes,
    gene_sets=["GO_Biological_Process_2023"],
    organism="mouse",
    cutoff=0.05,
    outdir=None,   
)

enr.results.head(20)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

res = enr.results.copy()

# --- choose top terms ---
TOP_K = 20
res = res.sort_values("Adjusted P-value").head(TOP_K).copy()

# --- metrics ---
# Fold enriched: prefer Odds Ratio if present
if "Odds Ratio" in res.columns:
    res["FoldEnriched"] = res["Odds Ratio"].astype(float)
elif "Combined Score" in res.columns:
    res["FoldEnriched"] = res["Combined Score"].astype(float)
else:
    raise ValueError("Couldn't find 'Odds Ratio' or 'Combined Score' in enr.results")

# Gene count from Overlap like "7/123"
res["GeneCount"] = res["Overlap"].str.split("/").str[0].astype(int)

# Color scale: -log10(FDR)
res["EnrichmentFDR"] = -np.log10(res["Adjusted P-value"].astype(float))

# Reverse so most significant appears at top (like your example)
res = res.sort_values("FoldEnriched", ascending=True)

# --- plot ---
fig, ax = plt.subplots(figsize=(9, 5))

y = np.arange(len(res))

# lollipop stems
ax.hlines(y=y, xmin=0, xmax=res["FoldEnriched"], linewidth=2)

# dots (size = gene count, color = enrichment)
sc = ax.scatter(
    res["FoldEnriched"],
    y,
    s=res["GeneCount"] * 12,       # tune this multiplier to taste
    c=res["EnrichmentFDR"],
)

ax.set_yticks(y)
ax.set_yticklabels(res["Term"])
ax.set_xlabel("FoldEnriched")
ax.set_title("Pathway Enrichment (Lollipop)\nSize = # Genes, Color = -log10(FDR)")

# colorbar
cbar = plt.colorbar(sc, ax=ax)
cbar.set_label("-log10(FDR)")

plt.tight_layout()
plt.show()

```
